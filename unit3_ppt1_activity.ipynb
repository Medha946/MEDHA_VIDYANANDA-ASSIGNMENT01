## Part 0 — Setup (given)

Run this cell as-is. It loads the camera log, the camera still, a frame renderer, and the audio (plus two
**reference recordings** from the security vendor's sound library: a known gunshot and a known glass break).

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import cbook
from scipy import ndimage, signal

np.random.seed(0)
plt.rcParams["axes.spines.top"] = False
plt.rcParams["axes.spines.right"] = False

FPS, FS = 30, 16_000
cctv = pd.read_csv("unit3_ppt1_cctv_log.csv")
img = plt.imread(cbook.get_sample_data("grace_hopper.jpg")).astype(float)
lum = 0.299 * img[..., 0] + 0.587 * img[..., 1] + 0.114 * img[..., 2]
night_still = 18 + 0.22 * lum + np.random.normal(0, 1.5, lum.shape)        # what the night camera delivers

FRAME_H, FRAME_W = 90, 160
def render_frame(row):
    yy, xx = np.mgrid[0:FRAME_H, 0:FRAME_W]
    base = row.mean_brightness * (0.75 + 0.25 * yy / FRAME_H)
    frame = np.dstack([base, base, base * 1.05]); frame[60:62, :, :] += 25
    if not np.isnan(row.car_x):
        frame[(np.abs(xx - row.car_x / 4) < 14) & (np.abs(yy - row.car_y / 4) < 6)] = [200, 40, 40]
    if not np.isnan(row.person_x):
        frame[((xx - row.person_x / 4) / 2.2) ** 2 + ((yy - row.person_y / 4) / 5) ** 2 <= 1] = [30, 90, 230]
    return np.clip(frame, 0, 255).astype(np.uint8)

def _clip(kind, seed):
    rng = np.random.default_rng(seed)
    t = np.arange(int(2.0 * FS)) / FS
    x = 0.004 * rng.normal(size=t.size)
    m = t >= 0.5
    if kind == "gunshot":                       # sharp broadband crack + low boom, nothing after
        x[m] += 0.9 * np.exp(-(t[m] - 0.5) / 0.02) * rng.normal(size=m.sum())
        x[m] += 0.5 * np.exp(-(t[m] - 0.5) / 0.25) * np.sin(2 * np.pi * 70 * (t[m] - 0.5))
    else:                                       # impact + lingering high-frequency tinkle
        x[m] += 0.9 * np.exp(-(t[m] - 0.5) / 0.03) * rng.normal(size=m.sum())
        for _ in range(40):
            t0 = 0.55 + rng.exponential(0.35); f0 = rng.uniform(3000, 7000)
            mk = (t >= t0) & (t < t0 + 0.06)
            x[mk] += 0.12 * np.exp(-(t[mk] - t0) / 0.015) * np.sin(2 * np.pi * f0 * (t[mk] - t0))
    return t, x

ref_gunshot = _clip("gunshot", 11)
ref_glass = _clip("glass", 12)
rng = np.random.default_rng(0)
t_audio = 70 + np.arange(10 * FS) / FS
x_audio = 0.02 * np.sin(2 * np.pi * 120 * t_audio) + 0.004 * rng.normal(size=t_audio.size)
_, unk = _clip("glass", 99)
x_audio[(t_audio >= 74.5) & (t_audio < 76.5)] += unk[: int(2 * FS)]           # the mystery sound

print(f"Camera log: {len(cctv):,} frames | still: {night_still.shape} | audio: {x_audio.size:,} samples at {FS} Hz")
cctv.head()
```

## Task 1 — The Storage Reality Check (Request #1)

**Real-world usage:** Storage budgets for camera fleets are decided from exactly these back-of-envelope numbers.

**Why this technique:** Uncompressed storage = **W × H × C × B** per frame (Slides 25–28), frames = **f × T**
(Slide 35), audio samples = **f_s × T** (Slide 55).

**TODO:**

2. Compute the number of frames ONE camera records in 24 h at 30 fps.
3. Compute the total uncompressed video for 400 cameras for 24 h, in **terabytes** (1 TB = 1024⁴ bytes).
4. Compute the same total if the cameras were downsampled to 640 × 360 and 10 fps. What percentage of the
   pixels per frame remain after the downsample?

```python
# TODO 1: one 1080p RGB frame, in MB
width, height, channels = 1920, 1080, 3
bytes_per_frame = width * height * channels
frame_mb = bytes_per_frame / 1_000_000

# TODO 2: frames per camera per day at 30 fps
FPS = 30
seconds_per_day = 24 * 60 * 60
frames_per_day = FPS * seconds_per_day

# TODO 3: 400 cameras x 24 h, uncompressed, in TB
num_cameras = 400
fleet_bytes = bytes_per_frame * frames_per_day * num_cameras
fleet_tb = fleet_bytes / (1024 ** 4)

# TODO 4: 640x360 at 10 fps
small_width, small_height, small_fps = 640, 360, 10
small_bytes_per_frame = small_width * small_height * channels
small_fleet_bytes = (
    small_bytes_per_frame
    * small_fps
    * seconds_per_day
    * num_cameras
)
fleet_tb_small = small_fleet_bytes / (1024 ** 4)

pct_pixels_remaining = (
    (small_width * small_height) / (width * height)
) * 100

print(f"One frame: {frame_mb:.2f} MB | frames/day: {frames_per_day:,}")
print(
    f"Fleet (1080p, 30 fps): {fleet_tb:,.1f} TB/day | "
    f"Fleet (360p, 10 fps): {fleet_tb_small:,.1f} TB/day"
)
print(
    f"Pixels per frame remaining after downsample: "
    f"{pct_pixels_remaining:.2f}%"
)
```

## Task 2 — Diagnose and Fix the Murky Night Image (Request #2)

**Real-world usage:** Before anyone interprets a night image, an analyst checks its histogram to see whether
the problem is exposure, contrast or clipping.

**Why this technique:** The **histogram** shows the intensity distribution (Slides 16–18); **Michelson
contrast** C = (L_max − L_min)/(L_max + L_min) quantifies it (Slide 15); a **contrast stretch** remaps the
occupied range onto 0–255.

**TODO:**
1. Plot `night_still` and its histogram (64 bins, range 0–255) side by side.
2. Compute its Michelson contrast using the 1st and 99th percentiles as L_min and L_max (robust to outliers).
3. Contrast-stretch it: map the 1st percentile → 0 and the 99th percentile → 255 (clip to 0–255).
4. Plot the stretched image + histogram, and print the new contrast.

```python
def michelson(lmax, lmin):
    denominator = lmax + lmin
    return (lmax - lmin) / denominator if denominator != 0 else 0

# TODO 2: robust Michelson contrast of night_still
lmin, lmax = np.percentile(night_still, [1, 99])
c_before = michelson(lmax, lmin)

# TODO 3: contrast-stretched version of night_still
stretched = np.clip(
    (night_still - lmin) / (lmax - lmin) * 255,
    0,
    255
)

# Calculate contrast after stretching using its 1st and 99th percentiles
new_min, new_max = np.percentile(stretched, [1, 99])
c_after = michelson(new_max, new_min)

# TODO 1 & 4: images and histograms before and after
fig, axes = plt.subplots(
    2, 2, figsize=(10, 8),
    gridspec_kw={"height_ratios": [2, 1]}
)

axes[0, 0].imshow(night_still, cmap="gray", vmin=0, vmax=255)
axes[0, 0].set_title("Original Night Image")
axes[0, 0].axis("off")

axes[1, 0].hist(
    night_still.ravel(), bins=64, range=(0, 255),
    color="gray"
)
axes[1, 0].set_title("Original Intensity Histogram")
axes[1, 0].set_xlabel("Intensity")
axes[1, 0].set_ylabel("Pixel Count")

axes[0, 1].imshow(stretched, cmap="gray", vmin=0, vmax=255)
axes[0, 1].set_title("Contrast-Stretched Image")
axes[0, 1].axis("off")

axes[1, 1].hist(
    stretched.ravel(), bins=64, range=(0, 255),
    color="gray"
)
axes[1, 1].set_title("Stretched Intensity Histogram")
axes[1, 1].set_xlabel("Intensity")
axes[1, 1].set_ylabel("Pixel Count")

plt.tight_layout()
plt.show()

print(f"Contrast before: {c_before:.3f} -> after: {c_after:.3f}")
```

**Reflection:** Describe the *shape* of the night image's histogram in the lecture's vocabulary (Slide 18 table).
Then explain why stretching made the image readable even though it added **no new information** — and why a
histogram alone could never have told you *what* is in the picture (Slide 18, "Same Histogram ≠ Same Image").

_Your answer:_
The original night image has a narrow histogram concentrated in a limited intensity range, indicating low contrast. Contrast stretching expands the occupied intensity range towards 0–255, making differences between dark and bright regions easier to see. It improves visibility but does not create new information; details that were never captured cannot be recovered by stretching. A histogram shows the distribution of pixel intensities, but it does not preserve the spatial arrangement of pixels, so it cannot tell us what objects or people appear in the image.

## Task 3 — From Two Minutes of Video to a One-Glance Timeline (Request #3)

**Real-world usage:** Security reviewers search for events; they don't watch footage linearly.

**Why this technique:** **Motion vectors → events → timeline + representative frames** (Slides 38–40, 45, 47).
Content-based keyframes beat uniform sampling when activity is sparse.

**TODO:**
1. Compute the person's speed in px/s as the magnitude of their motion vector over a 1-second window
   (`diff(FPS)` on x and y — single-frame differences are dominated by tracking jitter).
2. Find the time ranges where the person is **standing** (speed < 3 px/s) and **running** (speed > 25 px/s).
3. Detect the frames where `audio_level_db` first rises above −20 dB and where `mean_brightness` jumps by more
   than 40 in one frame.
4. Plot a timeline of `motion_energy` with a vertical marker + label for each event you found.
5. Show **one keyframe per event** (use `render_frame(cctv.iloc[i])`).

```python
# TODO 1: person speed (px/s), using a one-second window
person_x = cctv["person_x"]
person_y = cctv["person_y"]

dx = person_x - person_x.shift(FPS)
dy = person_y - person_y.shift(FPS)

speed = np.sqrt(dx**2 + dy**2)
speed = speed.where(
    person_x.notna()
    & person_y.notna()
    & person_x.shift(FPS).notna()
    & person_y.shift(FPS).notna()
)

cctv["person_speed"] = speed

# Helper function: print continuous time ranges for a Boolean condition
def print_time_ranges(mask, label):
    mask = pd.Series(mask, index=cctv.index).fillna(False).to_numpy()
    indices = np.flatnonzero(mask)

    print(f"\n{label}:")
    if len(indices) == 0:
        print("No matching frames found.")
        return

    # Split into continuous groups of frame indices
    groups = np.split(indices, np.where(np.diff(indices) > 1)[0] + 1)

    for group in groups:
        start = cctv.loc[group[0], "time_s"]
        end = cctv.loc[group[-1], "time_s"]
        print(f"{start:.2f} s to {end:.2f} s")

# TODO 2: standing and running ranges
standing = speed < 3
running = speed > 25

print_time_ranges(standing, "Standing (speed < 3 px/s)")
print_time_ranges(running, "Running (speed > 25 px/s)")

# TODO 3: first audio event and sudden lighting changes
audio_indices = np.flatnonzero(
    (cctv["audio_level_db"] > -20).fillna(False).to_numpy()
)
audio_event_idx = int(audio_indices[0]) if len(audio_indices) else None

brightness_change = cctv["mean_brightness"].diff()
lighting_indices = np.flatnonzero(
    (brightness_change.abs() > 40).fillna(False).to_numpy()
)
lighting_event_idx = int(lighting_indices[0]) if len(lighting_indices) else None

print("\nDetected events:")
if audio_event_idx is not None:
    print(
        "Audio first rises above -20 dB at:",
        f"{cctv.loc[audio_event_idx, 'time_s']:.2f} s"
    )
else:
    print("No audio threshold crossing detected.")

if lighting_event_idx is not None:
    print(
        "Brightness changes by more than 40 at:",
        f"{cctv.loc[lighting_event_idx, 'time_s']:.2f} s"
    )
else:
    print("No brightness jump greater than 40 detected.")

# TODO 4: annotated motion-energy timeline
fig, ax = plt.subplots(figsize=(12, 3))

ax.plot(
    cctv["time_s"], cctv["motion_energy"],
    linewidth=1, label="Motion energy"
)

if audio_event_idx is not None:
    t_audio_event = cctv.loc[audio_event_idx, "time_s"]
    ax.axvline(t_audio_event, color="tab:red", linestyle="--")
    ax.text(
        t_audio_event, 0.95, "Audio event",
        rotation=90, transform=ax.get_xaxis_transform(),
        va="top", ha="right", color="tab:red"
    )

if lighting_event_idx is not None:
    t_light_event = cctv.loc[lighting_event_idx, "time_s"]
    ax.axvline(t_light_event, color="tab:orange", linestyle="--")
    ax.text(
        t_light_event, 0.95, "Lighting event",
        rotation=90, transform=ax.get_xaxis_transform(),
        va="top", ha="right", color="tab:orange"
    )

ax.set_title("Parking Lot B: Motion Energy and Detected Events")
ax.set_xlabel("Time (seconds)")
ax.set_ylabel("Motion energy")
ax.legend()
ax.grid(alpha=0.2)

plt.tight_layout()
plt.show()

# TODO 5: one keyframe per detected event
keyframe_indices = []

# Add detected audio and lighting event frames
for idx in [audio_event_idx, lighting_event_idx]:
    if idx is not None:
        keyframe_indices.append(idx)

# Add the first frames where the person is standing or running
for mask in [standing, running]:
    valid_indices = np.flatnonzero(mask.fillna(False).to_numpy())
    if len(valid_indices):
        keyframe_indices.append(int(valid_indices[0]))

# Remove duplicates and display the frames
keyframe_indices = sorted(set(keyframe_indices))

if not keyframe_indices:
    print("No event keyframes were detected.")
else:
    fig, axes = plt.subplots(
        1, len(keyframe_indices),
        figsize=(4 * len(keyframe_indices), 3)
    )
    axes = np.atleast_1d(axes)

    for ax, idx in zip(axes, keyframe_indices):
        ax.imshow(render_frame(cctv.iloc[idx]))
        ax.set_title(f"Frame at {cctv.loc[idx, 'time_s']:.2f} s")
        ax.axis("off")

    plt.tight_layout()
    plt.show()
```

_Your answer:_

## Task 4 — Gunshot or Glass? (Request #4)

**Real-world usage:** Acoustic event classification in security systems starts with an analyst comparing
spectrogram signatures against known references.

**Why this technique:** A **waveform** shows *when* something loud happened but not *what* it was; a
**spectrogram** (Slides 56–58) reveals each sound's time–frequency signature.

**TODO:**
1. Plot the waveforms of `ref_gunshot`, `ref_glass` and the mystery segment of `x_audio` (74.5–76.5 s).
2. Plot their spectrograms (use `signal.spectrogram(x, fs=FS, nperseg=512, noverlap=384)`, show in dB,
   y-axis 0–8000 Hz).
3. Compute, for each clip, the fraction of spectral energy **above 3 kHz** in the 0.2–1.2 s after the impact.
4. Print a one-line verdict.

```python
mystery = x_audio[(t_audio >= 74.5) & (t_audio < 76.5)]

clips = {
    "Reference: gunshot": ref_gunshot[1],
    "Reference: glass break": ref_glass[1],
    "MYSTERY (74.5-76.5 s)": mystery
}

fig, axes = plt.subplots(2, 3, figsize=(15, 6))

for col, (name, samples) in enumerate(clips.items()):
    # TODO 1: waveform
    clip_time = np.arange(len(samples)) / FS
    axes[0, col].plot(clip_time, samples, linewidth=0.6)
    axes[0, col].set_title(name)
    axes[0, col].set_xlabel("Time (seconds)")
    axes[0, col].set_ylabel("Amplitude")

    # TODO 2: spectrogram
    freqs, times, power = signal.spectrogram(
        samples,
        fs=FS,
        nperseg=512,
        noverlap=384
    )

    power_db = 10 * np.log10(power + 1e-12)

    axes[1, col].pcolormesh(
        times, freqs, power_db,
        shading="auto",
        cmap="viridis"
    )
    axes[1, col].set_ylim(0, 8000)
    axes[1, col].set_xlabel("Time (seconds)")
    axes[1, col].set_ylabel("Frequency (Hz)")
    axes[1, col].set_title("Spectrogram (dB)")

plt.tight_layout()
plt.show()

# TODO 3: fraction of spectral energy above 3 kHz,
# from 0.2 to 1.2 seconds after the impact at 0.5 s
hf_fraction = {}

for name, samples in clips.items():
    freqs, times, power = signal.spectrogram(
        samples,
        fs=FS,
        nperseg=512,
        noverlap=384
    )

    time_mask = (times >= 0.7) & (times <= 1.7)
    high_freq_mask = freqs > 3000

    selected_power = power[:, time_mask]

    total_energy = np.sum(selected_power)
    high_energy = np.sum(selected_power[high_freq_mask, :])

    hf_fraction[name] = (
        high_energy / total_energy if total_energy > 0 else 0
    )

print("Fraction of spectral energy above 3 kHz:")
for name, fraction in hf_fraction.items():
    print(f"{name}: {fraction:.4f}")

# TODO 4: verdict
gunshot_fraction = hf_fraction["Reference: gunshot"]
glass_fraction = hf_fraction["Reference: glass break"]
mystery_fraction = hf_fraction["MYSTERY (74.5-76.5 s)"]

gunshot_distance = abs(mystery_fraction - gunshot_fraction)
glass_distance = abs(mystery_fraction - glass_fraction)

if gunshot_distance < glass_distance:
    verdict = "The mystery sound is more consistent with the gunshot reference."
else:
    verdict = "The mystery sound is more consistent with the glass-break reference."

print("\nVERDICT:", verdict)
```

**Reflection:** Look only at the three **waveforms**. Could you have confidently told the gunshot from the glass
break? What exactly does the spectrogram add (name the three dimensions from Slide 57), and which visual pattern
(horizontal band / vertical stripe / scattered high-frequency specks) decided your verdict?

_Your answer:_
The three waveforms show changes in amplitude over time, but the sharp initial impact can look similar for a gunshot and breaking glass, so the waveform alone is not sufficient for a confident classification. A spectrogram adds three dimensions: time, frequency and magnitude or energy. A gunshot tends to produce a sharp broadband vertical stripe followed by a decaying low-frequency component, while breaking glass may produce scattered high-frequency specks from lingering glass fragments. The high-frequency energy comparison and the visible spectrogram pattern help determine which reference the mystery sound more closely resembles.

## Task 5 — The One-Screen Briefing (Request #5)

**Real-world usage:** Incident briefings use a synchronized multimodal view so every claim is backed by two
modalities at the same timestamp.

**Why this technique:** **Image–video–audio integration** linked through a common time reference
(Slide 59), with **overview first** (timeline) and detail at the selected moment.

**TODO:** Build ONE figure with:
1. the rendered video frame at t = 75 s (left),
2. the full 120 s motion + audio timeline with a red marker at 75 s (top right),
3. the spectrogram of `x_audio` (70–80 s) with the same marker (bottom right),
4. a conclusion-style figure title (what happened, not "Incident Dashboard").

```python
t_sel = 75.0

# Find the CCTV frame closest to 75 seconds
frame_idx = (cctv["time_s"] - t_sel).abs().idxmin()
selected_row = cctv.loc[frame_idx]
selected_frame = render_frame(selected_row)

fig = plt.figure(figsize=(14, 6))
gs = fig.add_gridspec(
    2, 2,
    width_ratios=[1, 2],
    height_ratios=[1, 1.3]
)

# TODO 1: rendered video frame at t = 75 s (left)
ax_frame = fig.add_subplot(gs[:, 0])
ax_frame.imshow(selected_frame)
ax_frame.set_title(f"CCTV Keyframe at {selected_row['time_s']:.2f} s")
ax_frame.axis("off")

# TODO 2: full motion + audio timeline, top right
ax_timeline = fig.add_subplot(gs[0, 1])

ax_timeline.plot(
    cctv["time_s"],
    cctv["motion_energy"],
    label="Motion energy",
    color="tab:blue",
    linewidth=1
)
ax_timeline.set_xlabel("Time (seconds)")
ax_timeline.set_ylabel("Motion energy", color="tab:blue")
ax_timeline.tick_params(axis="y", labelcolor="tab:blue")

ax_audio_level = ax_timeline.twinx()
ax_audio_level.plot(
    cctv["time_s"],
    cctv["audio_level_db"],
    label="Audio level",
    color="tab:orange",
    alpha=0.65,
    linewidth=0.8
)
ax_audio_level.set_ylabel("Audio level (dB)", color="tab:orange")
ax_audio_level.tick_params(axis="y", labelcolor="tab:orange")

ax_timeline.axvline(
    t_sel, color="red", linestyle="--", label="75 s incident"
)
ax_timeline.set_title("120-Second Motion and Audio Overview")
ax_timeline.grid(alpha=0.2)

# Combine the legends from both axes
lines1, labels1 = ax_timeline.get_legend_handles_labels()
lines2, labels2 = ax_audio_level.get_legend_handles_labels()
ax_timeline.legend(lines1 + lines2, labels1 + labels2, loc="upper left")

# TODO 3: spectrogram of x_audio from 70-80 s, same marker
ax_spec = fig.add_subplot(gs[1, 1])

freqs, spec_times, power = signal.spectrogram(
    x_audio,
    fs=FS,
    nperseg=512,
    noverlap=384
)
power_db = 10 * np.log10(power + 1e-12)

absolute_times = 70 + spec_times

mesh = ax_spec.pcolormesh(
    absolute_times,
    freqs,
    power_db,
    shading="auto",
    cmap="viridis"
)

ax_spec.set_ylim(0, 8000)
ax_spec.set_xlim(70, 80)
ax_spec.axvline(
    t_sel, color="red", linestyle="--", label="75 s incident"
)
ax_spec.set_xlabel("Time (seconds)")
ax_spec.set_ylabel("Frequency (Hz)")
ax_spec.set_title("Audio Spectrogram (70–80 s)")
ax_spec.legend(loc="upper right")

fig.colorbar(mesh, ax=ax_spec, label="Power (dB)")

# TODO 4: conclusion-style title
fig.suptitle(
    "Parking Lot B: A Loud Impact Occurs Near 75 Seconds",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()
plt.show()
```
